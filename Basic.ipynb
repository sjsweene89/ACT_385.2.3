```python
# 1. Dictionary using curly brace notation
capitals_dict1 = {
    "California": "Los Angeles",
    "New York": "Albany",
    "Hawaii": "Honolulu",
    "Alaska": "Juneau",
    "Texas": "Austin"
}




# 2. Dictionary using the dict() constructor with a list of tuples
capitals_dict2 = dict([
    ("California", "Los Angeles"),
    ("New York", "Albany"),
    ("Hawaii", "Honolulu"),
    ("Alaska", "Juneau"),
    ("Texas", "Austin")
])

# 3. Check data types
print("Data type of capitals_dict1:", type(capitals_dict1))
print("Data type of capitals_dict2:", type(capitals_dict2))

# 4. Print dictionaries
print("\ncapitals_dict1:", capitals_dict1)
print("capitals_dict2:", capitals_dict2)


# Use capitals_dict1 from Problem 1
state_capitals = capitals_dict1.copy()

# 1. Retrieve the value associated with "California"
california_capital = state_capitals["California"]
print("Initial value for California:", california_capital)

# 2. Add a new key-value pair for Florida (Tallahassee)
state_capitals["Florida"] = "Tallahassee"

# 3. Update the value for "California" to "Sacramento"
state_capitals["California"] = "Sacramento"

# 4. Remove the key-value pair for "Alaska"
del state_capitals["Alaska"]

# Display updated dictionary
print("\nUpdated state_capitals dictionary:")
print(state_capitals)






playlist = {
    "Nipsey Hussle": "Double Up",
    "Beyonce": "speechless",
    "Billie Eilish": "Birds of a Feather",
    "Sabrina Carpenter": "Espresso",
    "Post Malone": "Chemical",
    "Dua Lipa": "Houdini"
}

# 2. For loop to print all artist names (keys)
print("--- Artists ---")
for artist in playlist.keys():
    print(artist)

# 3. For loop to print all song titles (values)
print("\n--- Songs ---")
for song in playlist.values():
    print(song)

# 4. For loop printing formatted sentence for each entry
print("\n--- Current Playlist Entries ---")
for artist, song in playlist.items():
    print(f'"{song}" by {artist} is in the current playlist.')

# 5. Remove the last key-value pair
removed_item = playlist.popitem()
print(f"\nRemoved last item: {removed_item}")

# 6. Add "Anti-Hero" by Taylor Swift
playlist["Taylor Swift"] = "Anti-Hero"

# 7. Update one song so that the title begins with "REMIX"
playlist["Sabrina Carpenter"] = f"REMIX - {playlist['Sabrina Carpenter']}"

# 8. Define and call a function to print all artists and songs
def display_playlist(catalog):
    """Iterates through a dictionary and prints all artists and songs."""
    print("\n--- Full Playlist Catalog ---")
    for artist, song in catalog.items():
        print(f"Artist: {artist} | Track: {song}")

display_playlist(playlist)
```
